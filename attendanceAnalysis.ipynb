```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_excel("aids_attendance.xlsx")
display(df.head())
```

```python
print("First 10 rows:")
display(df.head(10))

print("Shape:", df.shape)

print("Data types:")
print(df.dtypes)
```

```python
print(df.isnull().sum())

print("Never values in Last Present Date:")
print(
    df["Last Present Date"]
    .astype(str)
    .str.strip()
    .str.lower()
    .eq("never")
    .sum()
)
```

```python
df.columns = (
    df.columns
    .str.strip()
    .str.lower()
    .str.replace("%", "pct", regex=False)
    .str.replace(r"[^a-z0-9]+", "_", regex=True)
    .str.strip("_")
)

print(df.columns.tolist())
```

```python
df["attendance_pct"] = (
    df["attendance_pct"]
    .astype(str)
    .str.replace("%", "", regex=False)
    .astype(float)
)

df["last_present_date"] = pd.to_datetime(
    df["last_present_date"].replace("Never", pd.NA),
    errors="coerce"
)

display(df[["attendance_pct", "last_present_date"]].head())
```
